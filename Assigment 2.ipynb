## Q1 — List

```
roll = "1024170291"
digits = [int(d) for d in roll]

L = [d*10 for d in digits]
print(L)
```

```
L.append(55)
print(L)          # append() always adds the new value at the end
```

```
L.insert(2, 77)
print(L)          # insert() adds the value at the given index, shifting the rest right
```

```
L.remove(0)
print(L)          # remove() deletes the first matching value
```

```
L.pop()
print(L)          # pop() removes the last element by default
```

```
L.sort()
print(L)

L.sort(reverse=True)
print(L)
```

```
print(L[:3])
print(L[-3:])
```

```
avg = sum(L) / len(L)
above_avg = [x for x in L if x > avg]
print(above_avg)
```

## Q2 — Tuple

```
scores = tuple(L[:8])
print(scores)
```

```
highest = max(scores)
print(highest, scores.index(highest))

lowest = min(scores)
print(lowest, scores.count(lowest))
```

```
reversed_list = list(reversed(scores))
print(reversed_list)
# tuples are immutable, so there is no in-place reverse method like lists have
# reversed() just builds a new sequence instead of changing the original
```

```
val = int(input("enter a score: "))
if val in scores:
    print(scores.index(val))
else:
    print("not present")
```

```
try:
    scores[0] = 100
except TypeError as e:
    print(e)
# tuples don't support item assignment because they are immutable
# a list would allow scores[0] = 100 without any error
```

```
first, second, *rest = scores
print(first, second, rest)
```

## Q3 — Random List

```
import random

random.seed(1024170291)
nums = [random.randint(100, 900) for _ in range(100)]
```

```
odd = [x for x in nums if x % 2 != 0]
print(len(odd))

even = [x for x in nums if x % 2 == 0]
print(len(even))
```

```
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_count = 0
for x in nums:
    if is_prime(x):
        prime_count += 1
print(prime_count)

primes = [x for x in nums if is_prime(x)]
print(primes)
```

```
most_common = max(set(nums), key=nums.count)
print(most_common, nums.count(most_common))
```

## Q4 — Set

```
first8 = digits[:8]
A = {d * 7 for d in first8}
B = {d * 9 for d in first8}
print(A)
print(B)
```

```
print(A | B)
print(A & B)
```

```
print(A - B)
print(B - A)
# A - B gives values only in A, B - A gives values only in B
# symmetric_difference (A ^ B) combines both of those in one go

print(A ^ B)
```

```
print(A.issubset(B))
print(B.issuperset(A))
```

```
x = int(input("enter a value to remove from A: "))
A.discard(x)
print(A)
# discard() does nothing if x isn't in the set
# remove() would raise a KeyError instead, so discard() is safer here
```
